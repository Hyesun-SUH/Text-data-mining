```python

# 키워드를 입력하여 URL 형성
query = input('검색할 키워드를 입력하세요: ')
start_date = input('시작 날짜를 yyyy.mm.dd 형태로 입력하세요: ')
end_date = input('종료 날짜를 yyyy.mm.dd 형태로 입력하세요: ')

num_articles_per_page = 10  # 페이지당 기사 수
num_pages = int(input('추출할 페이지 수를 입력하세요: '))

# 데이터를 저장할 리스트 초기화
data = []

# 각 페이지에 대해 데이터 추출
for page in range(num_pages):
    start_index = page * num_articles_per_page + 1  # 페이지의 시작 인덱스 계산

    # URL 생성
    url = 'https://search.naver.com/search.naver?where=news&sm=tab_jum&query=' + \
          '%s' % query + "&start=" + str(start_index) + "&pd=3&ds=" + \
          '%s' % start_date + "&de=" + '%s' % end_date

    # HTML 문서 가져오기
    response = requests.get(url)
    html_text = response.text

    # HTML 파싱
    soup = bs(html_text, 'html.parser')

    # 뉴스 기사 제목 추출
    titles = soup.select('a.news_tit')

    # 추출한 제목을 데이터 리스트에 추가
    for title in titles:
        data.append(title.get_text())

# 데이터프레임 생성
df = pd.DataFrame(data, columns=['뉴스 제목'])

# CSV 파일로 저장
csv_filename = input('파일명을 입력하세요 : ')
df.to_csv(csv_filename, index=False, encoding='utf-8-sig')


```

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud
import pandas as pd
import re
from konlpy.tag import Okt
from collections import Counter

# 경로를 통해 CSV 파일 읽어오기
df = pd.read_csv("오염수.csv")

# C 열(뉴스 제목)의 데이터를 리스트로 가져오기
c_list = df['뉴스 제목'].tolist()

# 리스트를 하나의 벡터 자료로 합치기
vector = ''.join(c_list)

okt=Okt()

nouns = okt.nouns(vector)

# 길이가 1인 단어들을 정규 표현식을 이용하여 삭제
shortword = re.compile(r'\W*\b\w{1}\b')
nouns_str = ' '.join(nouns)  # 리스트 내의 모든 요소를 하나의 문자열로 합침
filtered_nouns_str = shortword.sub('', nouns_str)  # 정규 표현식을 이용하여 필터링

# 불용어 처리
stop_words = ["일본"]

# 불용어 제외한 단어들로 구성된 리스트 생성
filtered_words = [word for word in filtered_nouns_str.split() if word not in stop_words]

# 단어의 빈도수 계산
word_counts = Counter(filtered_words)

print(word_counts)
```

```python
#워드 클라우드 
file_path='C:/Users/seo hye sun/SNS텍스트데이터마이닝/'
wordcloud = WordCloud(font_path = file_path + 'NanumGothic/NanumGothicBold.ttf', background_color='white',colormap = "Accent", width=600, height=400).generate_from_frequencies(word_counts)


# 워드 클라우드 출력
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```
